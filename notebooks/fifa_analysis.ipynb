## 4. Data Quality & Cleaning

Before performing the exploratory analysis, the dataset is checked for missing values and duplicate records. This step helps ensure data consistency and improves the reliability of the analysis.

```python
# Check missing values
missing_values = df.isnull().sum()

missing_values[missing_values > 0].sort_values(ascending=False)
```

```python
# Calculate the percentage of missing values
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).round(2)

missing_percentage[missing_percentage > 0].sort_values(ascending=False)
```

```python
# Check duplicate records
duplicate_count = df.duplicated().sum()

print("Duplicate records:", duplicate_count)
```

### Data Cleaning Strategy

Based on the missing value analysis:

- `EMPRESTADO_DE_` contains 93.06% missing values. These values likely represent players who are not currently on loan, so they will be replaced with `"Not on loan"`.
- Missing values in `TIME` will be categorized as `"No Club"`.
- Missing values in `CONTRATACAO` and `VALIDADE_DO_CONTRATO` will be preserved, since replacing them without additional information could introduce inaccurate data.
- Rows with missing values in player attributes such as height, weight, agility, strength, and ball control represent a very small proportion of the dataset and can be removed when those attributes are required for analysis.
- No duplicate records were identified.

```python
# Create a copy for data cleaning
df_clean = df.copy()
```

```python
# Fill missing categorical values
df_clean["EMPRESTADO_DE_"] = df_clean["EMPRESTADO_DE_"].fillna("Not on loan")
df_clean["TIME"] = df_clean["TIME"].fillna("No Club")
```

```python
df_clean[["EMPRESTADO_DE_", "TIME"]].isnull().sum()
```

```python
# Remove rows with missing physical/technical attributes
df_clean = df_clean.dropna(
    subset=[
        "ALTURA",
        "PESO",
        "CONTROLE_DE_BOLA",
        "ARRANCADA",
        "AGILIDADE",
        "EQUILÍBRIO",
        "ENERGIA",
        "FORCA"
    ]
)
```

```python
print("Original dataset:", df.shape)
print("Clean dataset:", df_clean.shape)
```

```python
# Remove duplicate records
df_clean = df_clean.drop_duplicates()
```

```python
print("Original dataset:", df.shape)
print("Clean dataset:", df_clean.shape)
```

### Cleaning Results

After the data cleaning process:

- The original dataset contained 18,207 players.
- The cleaned dataset contains 18,159 players.
- Only 48 records (0.26%) were removed due to missing physical or technical attributes.
- Missing values in `EMPRESTADO_DE_` and `TIME` were handled using meaningful categorical replacements.
- No duplicate records were found.
- 99.74% of the original records were preserved after cleaning.

## 5. Nationality Analysis

This section analyzes the distribution of players by nationality and calculates the average player age for each nationality.

```python
# Number of unique nationalities
df_clean["NACIONALIDADE"].nunique()
```

```python
# Top 10 nationalities by number of players
top_nationalities = (
    df_clean["NACIONALIDADE"]
    .value_counts()
    .head(10)
)

top_nationalities
```

```python
# Average player age by nationality
average_age_by_nationality = (
    df_clean.groupby("NACIONALIDADE")["IDADE"]
    .mean()
    .round(2)
    .sort_values(ascending=False)
)

average_age_by_nationality
```

```python
# Nationality summary
nationality_summary = (
    df_clean.groupby("NACIONALIDADE")
    .agg(
        PLAYERS=("NOME", "count"),
        AVERAGE_AGE=("IDADE", "mean")
    )
    .sort_values("PLAYERS", ascending=False)
)

nationality_summary["AVERAGE_AGE"] = (
    nationality_summary["AVERAGE_AGE"].round(2)
)

nationality_summary.head(10)
```

## 6. Player Age Categorization

Players are categorized into two age groups using a custom Python function and the Pandas `.apply()` method.

For this analysis:
- Players aged 25 or younger are classified as **Young**.
- Players older than 25 are classified as **Veteran**.

```python
# Create age categories
def categorize_age(age):
    if age <= 25:
        return "Young"
    else:
        return "Veteran"

df_clean["AGE_CATEGORY"] = df_clean["IDADE"].apply(categorize_age)
```

```python
categorize_age(31)
```

```python
df_clean[
    ["NOME", "IDADE", "NACIONALIDADE", "AGE_CATEGORY"]
].head(10)
```

```python
age_category_count = df_clean["AGE_CATEGORY"].value_counts()

age_category_count
```

```python
age_category_percentage = (
    df_clean["AGE_CATEGORY"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

age_category_percentage
```

### Age Category Insights

- 10,071 players (55.46%) are classified as Young.

- Young players represent the majority of the dataset.

## 8. Player Performance Analysis

This section explores the relationship between player age, overall rating, and potential to identify performance patterns in the dataset.

```python
performance_by_category = (
    df_clean.groupby("AGE_CATEGORY")
    .agg(
        PLAYERS=("NOME", "count"),
        AVERAGE_AGE=("IDADE", "mean"),
        AVERAGE_OVERALL=("PONTUACAO_GERAL", "mean"),
        AVERAGE_POTENTIAL=("POTENCIAL", "mean")
    )
    .round(2)
)

performance_by_category
```

```python
top_players = (
    df_clean[
        ["NOME", "IDADE", "NACIONALIDADE", "TIME",
         "PONTUACAO_GERAL", "POTENCIAL"]
    ]
    .sort_values("PONTUACAO_GERAL", ascending=False)
    .head(10)
)

top_players
```

```python
top_young_potential = (
    df_clean[df_clean["AGE_CATEGORY"] == "Young"]
    [["NOME", "IDADE", "NACIONALIDADE", "TIME",
      "PONTUACAO_GERAL", "POTENCIAL"]]
    .sort_values("POTENCIAL", ascending=False)
    .head(10)
)

top_young_potential
```

### Performance Insights

- Veteran players have a higher average overall rating (69.17) compared to Young players (63.90).
- Young players show a higher average potential rating (72.69) compared to Veterans (69.61).
- The results suggest that Veteran players tend to have stronger current performance, while younger players offer greater potential for future development.
- This difference highlights the trade-off between current performance and future potential when evaluating players.

```python
df_clean["POTENTIAL_GROWTH"] = (
    df_clean["POTENCIAL"] - df_clean["PONTUACAO_GERAL"]
)
```

```python
high_growth_players = (
    df_clean[df_clean["AGE_CATEGORY"] == "Young"]
    [["NOME", "IDADE", "NACIONALIDADE", "TIME",
      "PONTUACAO_GERAL", "POTENCIAL", "POTENTIAL_GROWTH"]]
    .sort_values(
        ["POTENTIAL_GROWTH", "POTENCIAL"],
        ascending=[False, False]
    )
    .head(10)
)

high_growth_players
```

### Potential Growth Insights

- The `POTENTIAL_GROWTH` metric represents the difference between a player's potential rating and current overall rating.
- J. von Moos and D. Campbell show the highest potential growth in this analysis, with a 26-point difference.
- Several players aged 16 and 17 appear among those with the largest development margins.
- High potential growth does not necessarily indicate current high performance; instead, it highlights players whose potential rating is considerably higher than their current overall rating.
- This metric can support scouting analysis by identifying young players who may deserve further evaluation.

```python
performance_comparison = (
    df_clean.groupby("AGE_CATEGORY")
    [["PONTUACAO_GERAL", "POTENCIAL"]]
    .mean()
)

performance_comparison.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Average Overall Rating vs Potential by Age Category")
plt.xlabel("Age Category")
plt.ylabel("Average Rating")
plt.xticks(rotation=0)
plt.legend(["Overall Rating", "Potential"])
plt.tight_layout()
plt.show()
```

```python
growth_chart = (
    high_growth_players
    .set_index("NOME")["POTENTIAL_GROWTH"]
    .sort_values()
)

plt.figure(figsize=(10, 6))

growth_chart.plot(kind="barh")

plt.title("Top 10 Young Players by Potential Growth")
plt.xlabel("Potential Growth")
plt.ylabel("Player")
plt.tight_layout()
plt.show()
```
